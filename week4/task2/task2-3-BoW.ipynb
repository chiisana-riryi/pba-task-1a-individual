# 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
import string
import csv
from wordcloud import WordCloud

from sklearn.feature_extraction.text import CountVectorizer
```

## 2. Import Preprocessed data

### 2.1 Positive

```python
with open("../bow-data/positive.csv", newline="") as f:
    reader = csv.DictReader(f)
    data = list(reader)

positive_df = pd.DataFrame(data)

# convert dates to datetime
positive_df['at'] = pd.to_datetime(positive_df['at'])

positive_df.head()
```

### 2.2 Negative

```python
with open("../bow-data/negative.csv", newline="") as f:
    reader = csv.DictReader(f)
    data = list(reader)

negative_df = pd.DataFrame(data)

# convert dates to datetime
negative_df['at'] = pd.to_datetime(negative_df['at'])

negative_df.head()
```

## 3. BoW Analysis

```python

def generate_bow(content):
    out = {}
    text_tokens = content.split(" ")

    for token in text_tokens:
        if token in out:
            out[token] += 1
        else:
            out[token] = 1
    
    return out

print("Positive bags of words:")
for content in positive_df["content"]:
    positive_bow = generate_bow(content)
    print(positive_bow)
```

```python
print("Negative bags of words:")
for content in negative_df["content"]:
    negative_bow = generate_bow(content)
    print(negative_bow)
```

## 4. Using Scikit Sklearn

```python
def generate_bow_scikit(contents):
    cv = CountVectorizer()
    
    corpus = []
    for sentence in contents:
        corpus.append(sentence)

    independentFeatures = cv.fit_transform(corpus).toarray()
    return independentFeatures

positive_bow_scikit = generate_bow_scikit(positive_df["content"])

print(positive_bow_scikit)

negative_bow_scikit = generate_bow_scikit(negative_df["content"])

print(negative_bow_scikit)
```
