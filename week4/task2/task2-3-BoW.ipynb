# 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
import string
import csv
from wordcloud import WordCloud
```

## 2. Import Preprocessed data

### 2.1 Positive

```python
with open("../bow-data/positive.csv", newline="") as f:
    reader = csv.DictReader(f)
    data = list(reader)

positive_df = pd.DataFrame(data)

# convert dates to datetime
positive_df['at'] = pd.to_datetime(positive_df['at'])

positive_df.head()
```

### 2.2 Negative

```python
with open("../bow-data/negative.csv", newline="") as f:
    reader = csv.DictReader(f)
    data = list(reader)

negative_df = pd.DataFrame(data)

# convert dates to datetime
negative_df['at'] = pd.to_datetime(negative_df['at'])

negative_df.head()
```

## 3. BoW Analysis

```python

def generate_bow(contents):
    out = {}
    for text in contents:
        text_tokens = text.split(" ")

        for token in text_tokens:
            if token in out:
                out[token] += 1
            else:
                out[token] = 1
    
    return out

positive_bow = generate_bow(positive_df["content"])
print("Positive bag of words:")
print(positive_bow)

print("\n")

negative_bow = generate_bow(negative_df["content"])
print("Negative bag of words:")
print(negative_bow)
```
